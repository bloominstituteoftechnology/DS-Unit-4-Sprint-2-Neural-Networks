```python
%reload_ext tensorboard

import os
import datetime
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt

# It might be helpful to include the tensorbard code here.
logdir = os.path.join("logs", datetime.datetime.now().strftime("%Y%%m%d-%H%M%S"))
tensorboard_callback = tf.keras.callbacks.TensorBoard(logdir, histogram_freq=1)
```

```python
# I obviously need to grab the dataset.
# And since it comes with the Keras library, as mentioned in the instructions, I
# will import it into my notebook.

from tensorflow.keras.datasets import boston_housing
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()

# Pause here for a quick second: AKA - Data needs to be normalized
print("Shape:", X_train.shape)
print(X_train[:5])
# I will leave these commented out but come back to them if necessary.
# print("-----")
# print(y_train[:5])

# Now I can actually normalize the data.
from tensorflow.keras.utils import normalize

# Can't forget the X Matrices
X_train = normalize(X_train)
X_test = normalize(X_test)

# Next will create the same functions for the y vectors
y_train = normalize(y_train)[0]
y_test = normalize(y_test)[0]

# time to check my work.
print(X_train[:5])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, AlphaDropout, Dropout
from tensorflow.keras import regularizers

# Let's begin the modeling shall we?
model = Sequential()
model.add(Dense(128, activation='relu', input_dim=13))
model.add(Dropout(.2))
model.add(Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.01)))
model.add(Dropout(.2))
model.add(Dense(1, activation='relu'))

# This will create the early stopping function
from tensorflow.keras.callbacks import EarlyStopping
stop = EarlyStopping(monitor='val_mse', min_delta=0.001, patience=3)

# This function will compile the model
model.compile(loss='mse', optimizer='nadam', metrics=['mse'])
model.fit(X_train, y_train, epochs=125,
          validation_data=(X_test, y_test),
          callbacks=[tensorboard_callback])
```

## Use the Keras Library to build an image recognition network using the Fashion-MNIST dataset (also comes with keras)

- Load and preprocess the image data similar to how we preprocessed the MNIST data in class.
- Make sure to one-hot encode your category labels
- The number of nodes in your output layer should equal the number of classes you want to predict for Fashion-MNIST.
- Try different hyperparameters. What is the highest accuracy that you are able to achieve.
- Use the history object that is returned from model.fit to make graphs of the model's loss or train/validation accuracies by epoch. 
- Remember that neural networks fall prey to randomness so you may need to run your model multiple times (or use Cross Validation) in order to tell if a change to a hyperparameter is truly producing better results.

```python
from tensorflow.keras.layers import Flatten

# Again the Tensorboard as previously shown at the top of my notebook.
logdir = os.path.join("logs2", datetime.datetime.now().strftime("%Y%m%d-%H%M%S"))
tensorboard_callback = tf.keras.callbacks.TensorBoard(logdir, histogram_freq=1)

# This will create the MNIST Data
from tensorflow.keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Once again using the Early Stopping function
stop = EarlyStopping(monitor='val_accuracy', min_delta=0.01, patience=3)

# And now the Modeling
model_mnist = Sequential([
    Flatten(input_shape=(28,28)), # flatten the data coming in
    Dense(128, activation='relu', kernel_regularizer=regularizers.l1(0.01)),
    Dropout(.2),
    Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    Dropout(.2),
    Dense(128, activation='selu'),
    Dropout(.2),
    Dense(10, activation='softmax')
])

# Now I need to compile the model
model_mnist.compile(loss='sparse_categorical_crossentropy', optimizer='nadam', metrics=['accuracy'])

# Last but not least, I need to Fit the model
model_mnist.fit(X_train, y_train, epochs=125, 
          validation_data=(X_test, y_test),
          callbacks=[tensorboard_callback, stop]) # early stop callback
```
